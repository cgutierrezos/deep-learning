# Exponential Moving Average (EMA) en Python

Este notebook muestra un ejemplo de cómo calcular y visualizar un promedio móvil exponencial (EMA) sobre una serie de tiempo ruidosa, lo cual es fundamental para los optmizadores modernos.

## 1. Se generan los datos sintéticos

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Datos simulados
np.random.seed(42)
N = 100
x = np.linspace(0, 10, N)
y = np.sin(x) + np.random.normal(0, 0.5, N)

# Visualizar datos
plt.figure(figsize=(10, 6))
plt.plot(x, y, label='Datos originales', color='gray', alpha=0.6)
plt.xlabel('Tiempo')
plt.ylabel('Valor')
plt.title('Serie de tiempo ruidosa')
plt.legend()
plt.show()
```

## 2. Cálculo del Exponential Moving Average

El EMA se calcula con la fórmula:
$$ v_t = \alpha \cdot y_t + (1 - \alpha) \cdot v_{t-1} $$
donde $\alpha$ es el parámetro de suavizado (0 < $\alpha$ < 1).

```python
# Calcular EMA manualmente
alpha = 0.2  # parámetro de suavizado
v = np.zeros_like(y)
v[0] = y[0]
for t in range(1, N):
    v[t] = alpha * y[t] + (1 - alpha) * v[t -1]

# Graficar resultados
plt.figure(figsize=(10, 6))
plt.plot(x, y, label='Datos originales', color='gray', alpha=0.6)
plt.plot(x, v, label=f'EMA manual (alpha={alpha})', color='red', linewidth=2)
plt.xlabel('Tiempo')
plt.ylabel('Valor')
plt.title('Exponential Moving Average (EMA) - Implementación manual')
plt.legend()
plt.show()
```
